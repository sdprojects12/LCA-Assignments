```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load the csv dataset
df = pd.read_csv(r"C:\Users\Admin\Desktop\sample_data_assignment_07.csv")
```

```python
# Display The Dataset
print("Dataset: ")
print(df)
```

```python
plt.figure(figsize=(7,4.5))
plt.plot(df["Student"], df["Python"], marker="o")
plt.title("Python Marks of Students")
plt.xlabel("Student")
plt.ylabel("Python Marks")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(7,4.5))
plt.bar(df["Student"], df["Maths"])
plt.title("Maths Marks of Students")
plt.xlabel("Student")
plt.ylabel("Maths Marks")
plt.show()
```

```python
plt.figure(figsize=(7,4.5))
plt.scatter(df["Maths"], df["Python"], s=70)
plt.title("Maths Vs Python")
plt.xlabel("Maths")
plt.ylabel("Python Marks")
plt.grid(True)
plt.show()
```

```python
correlation = df[["Maths", "Python", "Attendance"]].corr()
print("Correlation Matrix: ")
print(correlation)
```

```python
plt.figure(figsize=(6.5,5))
sns.heatmap(correlation, annot=True, cmap="Blues", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.show()
```
