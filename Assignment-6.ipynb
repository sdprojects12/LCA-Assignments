```python
# Fill missing marks with the mean
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
```

```python
# Remove duplicate records
df = df.drop_duplicates()
```

```python
# Select students who scored above 80
result = df.loc[df["Marks"] > 80]
```
